```python
from pathlib import Path


student_id = "2025-0561"
folder_path = Path.home() / "Documents" / "Activity_5_Files"
files = list(folder_path.glob("*.txt"))

results = []


for file_path in files:
    text = file_path.read_text()
    
    # Split into lines and filter out empty ones
    lines = [line for line in text.splitlines() if line.strip()]
    words = text.split()
    characters = len(text.replace(" ", "").replace("\n", "")) # Ignore spaces/newlines
    
    # Calculations
    total_lines = len(lines)
    total_words = len(words)
    
    words_per_line = total_words / total_lines if total_lines > 0 else 0
    chars_per_word = characters / total_words if total_words > 0 else 0
    
    results.append({
        "Filename": file_path.name,
        "Lines": total_lines,
        "Words": total_words,
        "Chars": characters,
        "W/L": round(words_per_line, 2),
        "C/W": round(chars_per_word, 2)
    })


print(f"{'Filename':<25} | {'Lines':<5} | {'Words':<5} | {'Chars':<5} | {'W/L':<5} | {'C/W':<5}")
print("-" * 65)

for r in results:
    print(f"{r['Filename']:<25} | {r['Lines']:<5} | {r['Words']:<5} | {r['Chars']:<5} | {r['W/L']:<5} | {r['C/W']:<5}")


most_dense = max(results, key=lambda x: x['C/W'])
print(f"\nMost content-dense file: {most_dense['Filename']}")
```

```python
from pathlib import Path
import string


student_id = "2025-0561"
folder_path = Path.home() / "Documents" / "Activity_5_Files"
files = list(folder_path.glob("*.txt"))

stopwords = ["the", "is", "and", "to", "in", "a", "of", "it", "this", "for", "with"]
word_freq = {}


for file_path in files:
    
    text = file_path.read_text().lower()
    
   
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    words = text.split()
    
    for word in words:
    
        if word not in stopwords and not word.isdigit():
            word_freq[word] = word_freq.get(word, 0) + 1


sorted_words = sorted(word_freq.items(), key=lambda x: x[1], reverse=True)


print(f"--- Top Meaningful Words (Activity 5) ---")
for word, count in sorted_words[:10]: # Show top 10
    print(f"{word:<15}: {count}")
```

```python
import shutil
from pathlib import Path
from datetime import datetime


student_id = "2025-0561"
student_name = "Zyrah Angela Mae F. Balbin"


base_path = Path.home() / "Documents" / "Activity_5_Files"
backup_dir = base_path / f"backup_{student_id}"
backup_dir.mkdir(parents=True, exist_ok=True)


target_filename = f"intro_{student_id}.txt"
target_file = base_path / target_filename

if target_file.exists():
   
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
   
    
    backup_filename = f"{target_file.stem}_{student_id}_{timestamp}.txt"
    backup_path = backup_dir / backup_filename
    
   
    shutil.copy(target_file, backup_path)

    log_file = base_path / f"backup_log_{student_id}.txt"
    log_entry = (
        f"\n--- Backup Entry ---\n"
        f"Student: {student_name} ({student_id})\n"
        f"Original File: {target_filename}\n"
        f"Timestamp: {timestamp}\n"
        f"Size: {target_file.stat().st_size} bytes\n"
        f"Path: {backup_path}\n"
    )
    
    with log_file.open("a") as log:
        log.write(log_entry)
        
    print(f"Backup Successful!")
    print(f"File saved to: {backup_path}")
    print(f"Log updated at: {log_file}")

else:
    print(f"Error: Target file {target_filename} not found.")
```
